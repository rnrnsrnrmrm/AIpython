```python
country={"korea":"seoul","japan":"tokyo","usa":"washington dc"}

n=input("국가 혹은 수도를 입력: ")
if n in country:
  print(country[n])
elif n in country.values():
  for i,j in country.items():
    if j==n:
      print(i)
else:
  print("해당 국가나 수도를 찾을 수 없습니다.")
```
