```python
#과제 4-1, if문 미포함 반복문
s=0
for i in range(2,101,2):
  s+=i
print(s)
```

```python
#과제 4-1, if문 포함 반복문
s=0
for i in range(1,101):
  if i%2==0:
    s+=i
print(s)
```

```python
#2단부터 9단까지 구구단을 7 X 1 = 7 형식으로 출력하세요.
for i in range(2,10):
  for j in range(1,10):
    print(f"{i} X {j} = {i*j}")
  print("-------------------")
```

```python
#1에서 100까지 합 while문
s=0 ; i=1
while i<=100:
  s+=i
  i+=1
print(s)
```

```python
s=0 ; i=1
while True:
  s+=i
  i+=1
  if i>100:
    break
print(s)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하며 반복을 끝내고 합계를 출력하세요.
s=0
while True:
  n=int(input("양의 정수를 입력하세요"))
  s+=n
  if n==0:
    break
print(s)
```

```python
#1부터 20까지의 수 주 짝수만 담은 리스트를 만들어 출력
even=[]
for i in range(1,21):
  if i%2==0:
    even.append(i)
print(even)
even.pop(7)
print(even)
del even[5]
print(even)
```
