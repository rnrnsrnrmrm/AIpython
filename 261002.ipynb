```python
a=int(input("첫번째 수 : "))
b=int(input("두번째 수 : "))
opt=input("연산자 : ")

def func(a,b,opt):
  if opt=="+":
    print(f"{a} {opt} {b} = {a+b}")
  elif opt=="-":
    print(f"{a} {opt} {b} = {a-b}")
  elif opt=="*":
    print(f"{a} {opt} {b} = {a*b}")
  elif opt=="/":
    print(f"{a} {opt} {b} = {a/b}")
  else:
    print("다시쓰세요")

func(a,b,opt)

```
